#### 2. Compréhension des colonnes

Avant toute analyse, on classe les 46 colonnes selon leur **rôle** dans le projet. C'est essentiel : les relevés de consommation ne seront **pas connus** pour les nouveaux bâtiments, ils ne peuvent donc pas servir de features (sinon *fuite de données*).

| Groupe | Colonnes | Utilisation |
|---|---|---|
| Identifiants / texte libre | `OSEBuildingID`, `PropertyName`, `Address`, `TaxParcelIdentificationNumber` | Aucune (identifiants) |
| Constantes | `DataYear` (=2016), `City` (=Seattle), `State` (=WA), `Comments` (vide) | Aucune |
| Localisation | `ZipCode`, `CouncilDistrictCode`, `Neighborhood`, `Latitude`, `Longitude` | Features candidates |
| Structure | `YearBuilt`, `NumberofBuildings`, `NumberofFloors`, `PropertyGFATotal`, `PropertyGFAParking`, `PropertyGFABuilding(s)` | Features candidates |
| Usage | `BuildingType`, `PrimaryPropertyType`, `ListOfAllPropertyUseTypes`, `Largest/Second/ThirdLargestPropertyUseType(GFA)` | Filtrage + features candidates |
| Qualité de la donnée | `DefaultData`, `ComplianceStatus`, `Outlier` | **Filtrage des bâtiments peu fiables** |
| Relevés énergétiques | `SiteEnergyUse(kBtu)`, `SiteEnergyUseWN(kBtu)`, `SiteEUI…`, `SourceEUI…`, `Electricity…`, `NaturalGas…`, `SteamUse(kBtu)` | **Cibles candidates** / jamais en feature |
| Émissions | `TotalGHGEmissions`, `GHGEmissionsIntensity` | Cibles candidates / jamais en feature |
| Score ENERGY STAR | `ENERGYSTARScore`, `YearsENERGYSTARCertified` | Calculé *à partir* de la consommation → fuite, exclu |

Rappels utiles :
- **GFA** = *Gross Floor Area*, surface de plancher brute en pieds carrés (sf).
- **EUI** = *Energy Use Intensity* = consommation / surface (kBtu/sf).
- **WN** = *Weather Normalized*, consommation corrigée des variations météo.
- **Site** = énergie consommée sur place ; **Source** = énergie primaire (incluant pertes de production/transport).

```python
# Taux de valeurs manquantes par colonne
missing = (building_consumption.isna().mean() * 100).sort_values(ascending=False)
missing = missing[missing > 0]

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(missing.index[::-1], missing.values[::-1], color=BLUE, height=0.6)
for y, v in enumerate(missing.values[::-1]):
    ax.text(v + 1, y, f"{v:.1f} %", va="center", fontsize=8, color=GREY)
ax.set_xlabel("% de valeurs manquantes")
ax.set_xlim(0, 110)
ax.set_title("Taux de valeurs manquantes par colonne (jeu complet)")
plt.tight_layout(); plt.show()
```

**Lecture des valeurs manquantes — elles ne sont pas toutes des « trous » :**

- `Comments` est **entièrement vide** → colonne supprimée.
- `Outlier` est vide dans 99 % des cas : un vide signifie simplement *« le bâtiment n'est pas signalé comme outlier »*. C'est une information, pas un manque.
- `SecondLargestPropertyUseType(GFA)` / `ThirdLargest…` sont vides quand le bâtiment **n'a pas** de 2ᵉ ou 3ᵉ usage. La valeur logique est donc « aucun usage » et une surface de **0**.
- `YearsENERGYSTARCertified` est vide pour les bâtiments jamais certifiés ; `ENERGYSTARScore` n'est calculé que pour certains types d'usages éligibles. Ces deux colonnes sont de toute façon dérivées de la consommation (fuite) → exclues.
- Les relevés énergétiques ont ~9 manquants : ces bâtiments sont inutilisables pour l'entraînement (cible inconnue).

#### 3. Restriction au périmètre : bâtiments non destinés à l'habitation

On met en place un **journal de filtrage** pour tracer le nombre de lignes avant/après chaque étape.

```python
filter_log = []

def log_step(label, df_before, df_after):
    filter_log.append({
        "Étape": label,
        "Lignes avant": len(df_before),
        "Lignes après": len(df_after),
        "Lignes retirées": len(df_before) - len(df_after),
    })
    print(f"{label:<55} {len(df_before):>5} -> {len(df_after):>5}  (-{len(df_before) - len(df_after)})")

pd.crosstab(building_consumption["PrimaryPropertyType"], building_consumption["BuildingType"])
```

Le croisement `BuildingType` × `PrimaryPropertyType` montre que :
- les trois catégories **`Multifamily LR/MR/HR`** regroupent l'habitat collectif → à exclure ;
- quelques lignes classées *NonResidential* / *Campus* ont pourtant un usage principal **`Low-Rise Multifamily`** → exclues également ;
- `Residence Hall` (résidences étudiantes) est un usage **d'habitation** → exclu.

On conserve `Senior Care Community` : ce sont des établissements médicalisés (soins, restauration collective), dont le profil de consommation se rapproche davantage d'un bâtiment de services que d'un logement.

```python
df = building_consumption.copy()

before = df
df = df[~df["BuildingType"].str.startswith("Multifamily")]
log_step("BuildingType Multifamily (LR/MR/HR) exclus", before, df)

before = df
df = df[~df["PrimaryPropertyType"].isin(["Low-Rise Multifamily", "Residence Hall"])]
log_step("Usage principal résidentiel exclu", before, df)

# Uniformisation de libellés en doublon
df["PrimaryPropertyType"] = df["PrimaryPropertyType"].replace({"Office": "Small- and Mid-Sized Office"})
df["Neighborhood"] = df["Neighborhood"].str.upper().replace({"DELRIDGE NEIGHBORHOODS": "DELRIDGE"})

df["BuildingType"].value_counts()
```

Deux harmonisations de libellés au passage :
- les 3 bâtiments `Office` sont rattachés à `Small- and Mid-Sized Office` (catégorie la plus proche) ;
- `Neighborhood` mélangeait majuscules/minuscules (`North` vs `NORTH`) → passage en majuscules.

#### 4. Nettoyage : bâtiments aberrants, non conformes et incohérents

##### 4.1 Colonnes de qualité fournies par la ville
- `Outlier` : la ville a elle-même signalé certains bâtiments comme valeurs aberrantes (haute ou basse).
- `ComplianceStatus` : seuls les relevés **`Compliant`** sont validés. `Error - Correct Default Data` correspond exactement aux lignes `DefaultData = True` (valeurs par défaut saisies à la place des vraies) ; `Non-Compliant` et `Missing Data` sont non fiables.

```python
print(pd.crosstab(df["ComplianceStatus"], df["DefaultData"]), "\n")
print(df["Outlier"].value_counts(dropna=False))
```

```python
before = df
df = df[df["Outlier"].isna()]
log_step("Bâtiments signalés Outlier exclus", before, df)

before = df
df = df[df["ComplianceStatus"] == "Compliant"]
log_step("Relevés non conformes / données par défaut exclus", before, df)
```

##### 4.2 Cohérence des valeurs de features

On recherche des valeurs **physiquement impossibles ou contradictoires** :

```python
use_gfa_sum = df[["LargestPropertyUseTypeGFA", "SecondLargestPropertyUseTypeGFA",
                  "ThirdLargestPropertyUseTypeGFA"]].fillna(0).sum(axis=1)

checks = {
    "Consommation totale manquante": df["SiteEnergyUse(kBtu)"].isna(),
    "Consommation totale <= 0": df["SiteEnergyUse(kBtu)"] <= 0,
    "Émissions GES <= 0": df["TotalGHGEmissions"] <= 0,
    "Électricité négative": df["Electricity(kBtu)"] < 0,
    "Nombre de bâtiments = 0": df["NumberofBuildings"] == 0,
    "Nombre d'étages = 0": df["NumberofFloors"] == 0,
    "Surfaces des usages > 2x surface totale": use_gfa_sum > 2 * df["PropertyGFATotal"],
    "Somme GFA bâtiment + parking != GFA totale":
        (df["PropertyGFABuilding(s)"] + df["PropertyGFAParking"]) != df["PropertyGFATotal"],
}
pd.Series({k: int(v.sum()) for k, v in checks.items()}, name="Nb bâtiments").to_frame()
```

```python
# Le nombre d'étages : quelques valeurs extrêmes à examiner
df.nlargest(6, "NumberofFloors")[["PropertyName", "PrimaryPropertyType", "NumberofFloors", "PropertyGFATotal"]]
```

Une église de **99 étages** pour ~20 000 sf est manifestement une erreur de saisie (les autres valeurs élevées sont de vraies tours de bureaux du centre-ville, comme la Columbia Center à 76 étages).

**Décisions :**
- **Suppression** : cible manquante ou ≤ 0, émissions ≤ 0, électricité négative, l'église à 99 étages, et les bâtiments dont la somme des surfaces d'usage dépasse 2× la surface totale (surfaces contradictoires).
- **Correction plutôt que suppression** (pour ne pas perdre trop de bâtiments) :
  - `NumberofBuildings = 0` : une propriété déclarée possède forcément au moins un bâtiment → remplacé par **1** ;
  - `NumberofFloors = 0` : surtout des entrepôts/bâtiments bas ; remplacé par **1** (plain-pied).

```python
before = df
df = df[df["SiteEnergyUse(kBtu)"].notna() & (df["SiteEnergyUse(kBtu)"] > 0)]
log_step("Consommation manquante ou nulle exclue", before, df)

before = df
df = df[(df["TotalGHGEmissions"] > 0) & (df["Electricity(kBtu)"] >= 0)]
log_step("Émissions <= 0 / électricité négative exclues", before, df)

before = df
df = df[~((df["PrimaryPropertyType"] == "Worship Facility") & (df["NumberofFloors"] > 50))]
log_step("Nombre d'étages aberrant exclu", before, df)

before = df
use_gfa_sum = df[["LargestPropertyUseTypeGFA", "SecondLargestPropertyUseTypeGFA",
                  "ThirdLargestPropertyUseTypeGFA"]].fillna(0).sum(axis=1)
df = df[use_gfa_sum <= 2 * df["PropertyGFATotal"]]
log_step("Surfaces d'usage incohérentes exclues", before, df)

# Corrections
n_bld0, n_fl0 = (df["NumberofBuildings"] == 0).sum(), (df["NumberofFloors"] == 0).sum()
df.loc[df["NumberofBuildings"] == 0, "NumberofBuildings"] = 1
df.loc[df["NumberofFloors"] == 0, "NumberofFloors"] = 1
print(f"\nCorrigés : NumberofBuildings 0->1 : {n_bld0} | NumberofFloors 0->1 : {n_fl0}")
```

#### 5. Choix de la variable cible

Plusieurs colonnes peuvent servir de cible :

| Candidate | Pour | Contre |
|---|---|---|
| `SiteEnergyUse(kBtu)` | Consommation **totale** réelle : c'est exactement ce que demande la ville | Très asymétrique |
| `SiteEnergyUseWN(kBtu)` | Corrigée de la météo | Valeurs à 0 pour des bâtiments qui consomment (correction non calculée) |
| `SiteEUI(kBtu/sf)` | Indépendante de la taille | Intensité, pas consommation totale ; la ville veut la consommation |
| `TotalGHGEmissions` | Objectif carbone | Dépend surtout du **mix énergétique** (gaz, vapeur), non connu à partir des seules données structurelles |

```python
targets = ["SiteEnergyUse(kBtu)", "SiteEnergyUseWN(kBtu)", "SiteEUI(kBtu/sf)", "TotalGHGEmissions"]
print("Valeurs nulles par candidate :")
print((df[targets] == 0).sum(), "\n")
print("Corrélation (Spearman) entre candidates :")
df[targets].corr(method="spearman")
```

✅ **Cible retenue pour toute la suite du projet : `SiteEnergyUse(kBtu)`** — consommation totale d'énergie du site.

- Elle correspond directement à la demande métier (« prédire la consommation totale d'énergie ») ;
- elle n'a plus de valeur nulle ou manquante après nettoyage, contrairement à la version WN ;
- elle est fortement liée aux émissions (Spearman élevé), donc une bonne prédiction de consommation éclaire aussi l'enjeu carbone.

Les autres relevés énergétiques (EUI, électricité, gaz, vapeur, émissions, ENERGY STAR) seront **exclus des features** pour éviter toute fuite de données.

```python
TARGET = "SiteEnergyUse(kBtu)"

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(df[TARGET] / 1e6, bins=60, color=BLUE)
axes[0].set_title("Consommation totale (échelle linéaire)")
axes[0].set_xlabel("SiteEnergyUse (millions de kBtu)"); axes[0].set_ylabel("Nb bâtiments")

axes[1].hist(np.log10(df[TARGET]), bins=40, color=BLUE)
axes[1].set_title("Consommation totale (échelle log10)")
axes[1].set_xlabel("log10(SiteEnergyUse en kBtu)")
plt.tight_layout(); plt.show()

print(f"Skewness brute : {df[TARGET].skew():.1f} | skewness log : {np.log10(df[TARGET]).skew():.2f}")
```

La distribution brute est **extrêmement asymétrique** (quelques très gros bâtiments : hôpitaux, université, tours). En échelle logarithmique, elle devient proche d'une **gaussienne**.
→ Pour la modélisation, il sera pertinent de prédire **log(cible)** (ou d'utiliser une transformation équivalente), pour que l'erreur ne soit pas dominée par une poignée de très gros bâtiments.

```python
df["log_SiteEnergyUse"] = np.log10(df[TARGET])
```

##### 7.1 Quanti × quanti

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
axes[0].scatter(np.log10(df["PropertyGFATotal"]), df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[0].set_xlabel("log10(Surface totale, sf)"); axes[0].set_ylabel("log10(Consommation, kBtu)")
axes[0].set_title("Consommation vs surface")

axes[1].scatter(df["NumberofFloors"], df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[1].set_xlabel("Nombre d'étages"); axes[1].set_title("Consommation vs nombre d'étages")

axes[2].scatter(df["YearBuilt"], df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[2].set_xlabel("Année de construction"); axes[2].set_title("Consommation vs année de construction")
plt.tight_layout(); plt.show()
```

```python
corr_cols = ["log_SiteEnergyUse", "PropertyGFATotal", "PropertyGFABuilding(s)", "PropertyGFAParking",
             "LargestPropertyUseTypeGFA", "NumberofFloors", "NumberofBuildings", "BuildingAge",
             "Latitude", "Longitude"]
corr = df[corr_cols].corr(method="spearman")

fig, ax = plt.subplots(figsize=(9, 7))
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1,
            annot_kws={"size": 8}, linewidths=1, linecolor="white", cbar_kws={"shrink": 0.7}, ax=ax)
ax.set_title("Corrélations de Spearman (features structurelles et log-cible)")
plt.tight_layout(); plt.show()
```

- La **surface** est de très loin le premier déterminant de la consommation (Spearman ≈ 0,74) : relation quasi linéaire en log-log.
- Les différentes surfaces sont **fortement corrélées entre elles** (multicolinéarité) : il faudra en choisir une ou deux pour les modèles linéaires.
- Le nombre d'étages est modérément corrélé (0,45), surtout via la taille.
- L'**âge** est négativement corrélé (−0,28) : les bâtiments récents (années 1990-2010) consomment davantage, mais ils sont aussi plus grands (corrélation âge/surface −0,28). C'est un effet de taille plus qu'un effet d'efficacité.
- La **latitude/longitude** brutes n'ont quasiment aucun lien linéaire avec la cible.

```python
# Pour séparer l'effet "taille" de l'effet "usage" : intensité énergétique (EUI) par usage
# (EUI utilisée ici uniquement pour l'analyse, pas comme feature)
order_eui = df.groupby("PrimaryPropertyType")["SiteEUI(kBtu/sf)"].median().sort_values(ascending=False).index
fig, ax = plt.subplots(figsize=(10, 7))
sns.boxplot(data=df, y="PrimaryPropertyType", x="SiteEUI(kBtu/sf)", order=order_eui,
            color=ORANGE, fliersize=2, linewidth=1, ax=ax, boxprops={"alpha": 0.8})
ax.set_xscale("log"); ax.set_xlabel("Intensité énergétique SiteEUI (kBtu/sf, échelle log)"); ax.set_ylabel("")
ax.set_title("Intensité énergétique par m² selon l'usage principal")
ax.tick_params(axis="y", labelsize=8)
plt.tight_layout(); plt.show()
```

- L'**usage** influence fortement la consommation, même à surface égale : hôpitaux, laboratoires, supermarchés et restaurants consomment **5 à 15 fois plus par unité de surface** (médiane ≈ 220-250 kBtu/sf contre 15-30) que les entrepôts, self-storage ou lieux de culte.
- La consommation totale élevée des universités et grands bureaux s'explique surtout par leur **taille**, alors que celle des hôpitaux, supermarchés et laboratoires s'explique par leur **intensité**.
→ L'usage principal (et la répartition des surfaces par usage) sera une feature clé, complémentaire de la surface.

### Feature Engineering

A réaliser : Enrichir le jeu de données actuel avec de nouvelles features issues de celles existantes. 

Le feature engineering crée de nouvelles variables à partir des colonnes **structurelles** du jeu nettoyé. Deux règles guident toute cette partie :

1. **Pas de fuite de données.** Une feature doit pouvoir être calculée pour un bâtiment dont on n'a **pas encore mesuré** la consommation. On n'utilise donc jamais les valeurs des relevés (consommation, EUI, électricité, gaz, vapeur, émissions, ENERGY STAR).
2. **Pas d'apprentissage sur tout le jeu avant le split.** Toutes les features ci-dessous sont calculées **ligne par ligne**, à partir des seules colonnes du bâtiment, avec des règles fixées à l'avance (seuils, listes, point de référence). Elles peuvent donc être créées avant la séparation train/test sans fuite. Les transformations qui *apprennent* sur les données (regroupement des modalités rares selon leur fréquence, standardisation, encodage) seront faites dans un `Pipeline` sklearn, ajusté uniquement sur le jeu d'entraînement, dans la partie suivante.

Les features couvrent quatre catégories d'information : **temporalité**, **structure**, **usages** et **localisation**, plus des indicateurs structurels sur les **sources d'énergie** présentes.

```python
df_fe = building_consumption_clean.copy()
n_cols_initial = df_fe.shape[1]

# Colonnes issues des relevés énergétiques : interdites en feature (connues seulement après mesure)
LEAKAGE_COLS = [
    "SiteEUI(kBtu/sf)", "SiteEUIWN(kBtu/sf)", "SourceEUI(kBtu/sf)", "SourceEUIWN(kBtu/sf)",
    "SiteEnergyUse(kBtu)", "SiteEnergyUseWN(kBtu)", "SteamUse(kBtu)",
    "Electricity(kWh)", "Electricity(kBtu)", "NaturalGas(therms)", "NaturalGas(kBtu)",
    "TotalGHGEmissions", "GHGEmissionsIntensity", "log_SiteEnergyUse",
]
new_features = {}   # nom de la feature -> catégorie, pour le récapitulatif final
```

#### 1. Temporalité

- `BuildingAge` (déjà créée pendant l'analyse exploratoire) : âge au moment du relevé, `2016 - YearBuilt`. On utilise l'année des données (2016) et non l'année en cours, sinon l'âge n'aurait pas de sens par rapport à la consommation mesurée.
- `ConstructionPeriod` : période de construction en 4 tranches. Les techniques de construction et d'isolation évoluent par grandes époques. Une variable par tranche aide surtout les modèles linéaires, qui ne captent pas seuls un effet non linéaire de l'âge. Les bornes sont fixées **à l'avance** à partir de repères historiques (après-guerre, chocs pétroliers des années 1970 et premières réglementations thermiques, années 2000), et non par quantiles calculés sur les données.

```python
df_fe["BuildingAge"] = 2016 - df_fe["YearBuilt"]

df_fe["ConstructionPeriod"] = pd.cut(
    df_fe["YearBuilt"],
    bins=[0, 1945, 1979, 1999, 2016],
    labels=["Avant 1946", "1946-1979", "1980-1999", "2000 et après"],
).astype(str)

new_features.update({"BuildingAge": "Temporalité", "ConstructionPeriod": "Temporalité"})
df_fe["ConstructionPeriod"].value_counts().reindex(["Avant 1946", "1946-1979", "1980-1999", "2000 et après"]).to_frame("Nb bâtiments")
```

#### 2. Structure du bâtiment

- `log_GFATotal`, `log_GFABuilding` : logarithmes des surfaces. L'analyse exploratoire a montré que la surface est le premier facteur de consommation et que la relation est **linéaire en log-log**. Le log rend donc cette relation directement exploitable par un modèle linéaire.
- `GFAPerFloor` : surface moyenne par étage (forme du bâtiment). Un bâtiment large et bas (entrepôt) et une tour fine n'ont pas la même exposition ni les mêmes équipements (ascenseurs, ventilation).
- `GFAPerBuilding` : surface moyenne par bâtiment, qui distingue un campus de nombreux petits bâtiments d'un bâtiment unique.
- `ParkingShare` : part de la surface totale occupée par du parking, très peu consommateur par m².
- `HasParking`, `IsMultiBuilding` : indicateurs binaires simples (78 % des propriétés n'ont pas de parking, 97 % n'ont qu'un bâtiment).

`GFAPerFloor` est très asymétrique : un campus déclaré avec un seul étage atteint plus de 9 millions de sf par étage. On ajoute donc aussi sa version log, `log_GFAPerFloor`.

Les divisions sont sans risque : `NumberofFloors` et `NumberofBuildings` ont été corrigés à 1 minimum pendant le nettoyage. On le vérifie quand même ci-dessous (`assert`).

```python
assert (df_fe["NumberofFloors"] >= 1).all() and (df_fe["NumberofBuildings"] >= 1).all()

df_fe["log_GFATotal"] = np.log10(df_fe["PropertyGFATotal"])
df_fe["log_GFABuilding"] = np.log10(df_fe["PropertyGFABuilding(s)"])
df_fe["GFAPerFloor"] = df_fe["PropertyGFABuilding(s)"] / df_fe["NumberofFloors"]
df_fe["log_GFAPerFloor"] = np.log10(df_fe["GFAPerFloor"])
df_fe["GFAPerBuilding"] = df_fe["PropertyGFABuilding(s)"] / df_fe["NumberofBuildings"]
df_fe["ParkingShare"] = df_fe["PropertyGFAParking"] / df_fe["PropertyGFATotal"]
df_fe["HasParking"] = (df_fe["PropertyGFAParking"] > 0).astype(int)
df_fe["IsMultiBuilding"] = (df_fe["NumberofBuildings"] > 1).astype(int)

struct_feats = ["log_GFATotal", "log_GFABuilding", "GFAPerFloor", "log_GFAPerFloor",
                "GFAPerBuilding", "ParkingShare", "HasParking", "IsMultiBuilding"]
new_features.update({f: "Structure" for f in struct_feats})
df_fe[struct_feats].describe().T
```

`ParkingShare` reste bien entre 0 et 1 : la surface totale vaut exactement bâtiment + parking (vérifié pendant le nettoyage).

#### 3. Usages du bâtiment

##### 3.1 Nombre d'usages et bâtiments multi-usages

`ListOfAllPropertyUseTypes` liste tous les usages séparés par des virgules. **Piège** : certains noms d'usage contiennent eux-mêmes des virgules entre parenthèses, par exemple `Personal Services (Health/Beauty, Dry Cleaning, etc)`. Un simple `split(",")` compterait donc trop d'usages pour ces bâtiments. On découpe uniquement sur les virgules **hors parenthèses** grâce à une expression régulière.

```python
import re

# virgule suivie d'espaces, sauf si on se trouve à l'intérieur d'une parenthèse
USE_SEP = re.compile(r",\s*(?![^()]*\))")

def split_uses(use_list):
    return [u.strip() for u in USE_SEP.split(use_list) if u.strip()]

use_lists = df_fe["ListOfAllPropertyUseTypes"].apply(split_uses)

# Contrôle : comparaison avec le découpage naïf
naive_count = df_fe["ListOfAllPropertyUseTypes"].str.split(",").str.len()
print(f"Bâtiments dont le nombre d'usages serait faux avec split(','): {(naive_count != use_lists.str.len()).sum()}")

df_fe["NumUseTypes"] = use_lists.str.len()
df_fe["IsMultiUse"] = (df_fe["NumUseTypes"] > 1).astype(int)
new_features.update({"NumUseTypes": "Usages", "IsMultiUse": "Usages"})

df_fe["NumUseTypes"].value_counts().sort_index().to_frame("Nb bâtiments").T
```

##### 3.2 Répartition des surfaces entre usages (proportions)

On calcule la part de surface des 1er, 2e et 3e usages. Un bâtiment à 100 % bureaux ne se comporte pas comme un bâtiment 60 % bureaux et 40 % restaurant.

Le dénominateur est la **somme des surfaces des trois usages**, et non `PropertyGFATotal` : le nettoyage a montré que la somme des surfaces d'usage peut dépasser la surface totale (jusqu'à 2×). Diviser par la surface totale donnerait des « parts » supérieures à 1. Avec ce dénominateur, les trois parts sont comprises entre 0 et 1 et leur somme vaut 1.

```python
use_gfa_cols = ["LargestPropertyUseTypeGFA", "SecondLargestPropertyUseTypeGFA", "ThirdLargestPropertyUseTypeGFA"]
use_gfa_total = df_fe[use_gfa_cols].sum(axis=1)
assert (use_gfa_total > 0).all()

df_fe["LargestUseShare"] = df_fe["LargestPropertyUseTypeGFA"] / use_gfa_total
df_fe["SecondUseShare"] = df_fe["SecondLargestPropertyUseTypeGFA"] / use_gfa_total
df_fe["ThirdUseShare"] = df_fe["ThirdLargestPropertyUseTypeGFA"] / use_gfa_total

share_feats = ["LargestUseShare", "SecondUseShare", "ThirdUseShare"]
new_features.update({f: "Usages" for f in share_feats})
df_fe[share_feats].describe().T
```

##### 3.3 Usages énergivores

Certains usages consomment beaucoup plus par m² que les autres : data centers, hôpitaux, laboratoires, restauration et commerces alimentaires (réfrigération, cuisson), entrepôts frigorifiques.

**Point de vigilance** : la liste est fixée à partir de **connaissances métier** (catégories d'usage reconnues comme les plus intensives dans les références sectorielles, par exemple ENERGY STAR), et non en choisissant les usages qui maximisent la consommation dans *notre* jeu de données. Sélectionner les catégories d'après la cible observée sur tout le jeu (test inclus) serait une forme subtile de fuite.

Deux familles de features :
- `HighIntensityShare` : part de la surface (3 premiers usages) occupée par un usage énergivore ;
- des indicateurs binaires de **présence**, calculés sur la **liste complète** des usages (qui va au-delà des 3 premiers) : `HasDataCenter`, `HasFoodService`, `HasHealthcare`, `HasLab`, `HasRefrigeratedStorage`.

```python
USE_GROUPS = {
    "HasDataCenter": ["Data Center"],
    "HasFoodService": ["Restaurant", "Other - Restaurant/Bar", "Food Service", "Food Sales",
                       "Supermarket/Grocery Store", "Convenience Store without Gas Station",
                       "Convenience Store with Gas Station", "Bar/Nightclub", "Fast Food Restaurant"],
    "HasHealthcare": ["Hospital (General Medical & Surgical)", "Other/Specialty Hospital",
                      "Urgent Care/Clinic/Other Outpatient", "Outpatient Rehabilitation/Physical Therapy",
                      "Ambulatory Surgical Center"],
    "HasLab": ["Laboratory"],
    "HasRefrigeratedStorage": ["Refrigerated Warehouse"],
}
HIGH_INTENSITY_USES = set().union(*USE_GROUPS.values())

for feat, uses in USE_GROUPS.items():
    df_fe[feat] = use_lists.apply(lambda lst: int(any(u in uses for u in lst)))

high_gfa = sum(
    df_fe[f"{rank}PropertyUseTypeGFA"].where(df_fe[f"{rank}PropertyUseType"].isin(HIGH_INTENSITY_USES), 0)
    for rank in ["Largest", "SecondLargest", "ThirdLargest"]
)
df_fe["HighIntensityShare"] = high_gfa / use_gfa_total

# Contrôle : chaque nom de la liste doit exister dans les données (sinon faute de frappe silencieuse)
all_uses = set(u for lst in use_lists for u in lst)
print("Usages de la liste absents des données (normal s'ils sont simplement rares) :",
      sorted(HIGH_INTENSITY_USES - all_uses))

intensity_feats = list(USE_GROUPS) + ["HighIntensityShare"]
new_features.update({f: "Usages" for f in intensity_feats})
df_fe[intensity_feats].agg(["mean", "max"]).T.rename(columns={"mean": "Part / moyenne", "max": "Max"})
```

Les noms absents de la liste ci-dessus ne posent pas de problème : la liste reste valable pour de futurs bâtiments qui auraient ces usages. Le contrôle sert surtout à vérifier que les noms **présents** dans les données sont bien orthographiés (une faute de frappe donnerait silencieusement une colonne de 0).

##### 3.4 Regroupement des usages en familles (réduction de la cardinalité)

`LargestPropertyUseType` compte **56 modalités**, dont beaucoup ne concernent que 1 à 5 bâtiments : un encodage one-hot créerait 56 colonnes, la plupart presque vides. `PrimaryPropertyType` (19 modalités) contient une catégorie fourre-tout `Other` de 239 bâtiments, alors que `LargestPropertyUseType` précise souvent ce qu'elle recouvre (loisirs, salles de spectacle, industrie…).

On crée `UseFamily` : l'usage principal détaillé regroupé en **10 familles** au comportement énergétique proche. Le regroupement repose sur un dictionnaire fixé à la main (sens métier), et non sur la fréquence ou la consommation observée. Il peut donc être fait avant le split. Un usage inconnu (nouveau bâtiment) tombe dans la famille `Autre`.

```python
USE_FAMILY = {
    "Bureaux": ["Office", "Small- and Mid-Sized Office", "Financial Office", "Bank Branch", "Courthouse"],
    "Commerce": ["Retail Store", "Strip Mall", "Other - Mall", "Lifestyle Center", "Automobile Dealership",
                 "Wholesale Club/Supercenter", "Enclosed Mall",
                 "Personal Services (Health/Beauty, Dry Cleaning, etc)",
                 "Repair Services (Vehicle, Shoe, Locksmith, etc)"],
    "Alimentaire & restauration": ["Supermarket/Grocery Store", "Restaurant", "Other - Restaurant/Bar",
                                   "Food Service", "Food Sales"],
    "Entrepôt & stockage": ["Non-Refrigerated Warehouse", "Distribution Center", "Self-Storage Facility", "Parking"],
    "Industrie, froid & data": ["Refrigerated Warehouse", "Manufacturing/Industrial Plant", "Data Center",
                                "Other - Utility"],
    "Santé": ["Hospital (General Medical & Surgical)", "Other/Specialty Hospital", "Medical Office",
              "Urgent Care/Clinic/Other Outpatient", "Laboratory", "Senior Care Community",
              "Residential Care Facility"],
    "Enseignement": ["K-12 School", "College/University", "Other - Education", "Adult Education",
                     "Pre-school/Daycare", "Library"],
    "Hébergement": ["Hotel", "Other - Lodging/Residential", "Multifamily Housing",
                    "Residence Hall/Dormitory", "Prison/Incarceration"],
    "Culte, culture & loisirs": ["Worship Facility", "Other - Recreation", "Other - Entertainment/Public Assembly",
                                 "Social/Meeting Hall", "Fitness Center/Health Club/Gym", "Museum",
                                 "Performing Arts", "Movie Theater"],
    "Autre": ["Other", "Other - Services", "Other - Public Services", "Police Station", "Fire Station"],
}
use_to_family = {use: fam for fam, uses in USE_FAMILY.items() for use in uses}

df_fe["UseFamily"] = df_fe["LargestPropertyUseType"].map(use_to_family).fillna("Autre")

unmapped = sorted(set(df_fe["LargestPropertyUseType"]) - set(use_to_family))
print("Usages non rattachés à une famille :", unmapped if unmapped else "aucun")
new_features["UseFamily"] = "Usages"
df_fe["UseFamily"].value_counts().to_frame("Nb bâtiments")
```

#### 4. Localisation

L'analyse exploratoire a montré que la latitude et la longitude **brutes** n'ont presque aucun lien avec la consommation, alors que **Downtown** se distingue nettement. On résume donc la position par une information plus parlante pour un modèle :

- `DistanceToDowntown_km` : distance (formule de haversine) entre le bâtiment et un point de référence **fixe** du centre-ville (Westlake Center, 47.6114 N, -122.3374 O). Un point fixe ne dépend pas des données, donc aucune fuite.
- `IsDowntown` : indicateur binaire du quartier `DOWNTOWN`.

`Neighborhood` (13 modalités) a déjà été nettoyé (casse harmonisée) et pourra être encodé tel quel. `ZipCode` (48 modalités, 13 manquants) et `CouncilDistrictCode` apportent une information géographique redondante avec le quartier et la distance : ils seront écartés à l'étape de préparation.

```python
DOWNTOWN_LAT, DOWNTOWN_LON = 47.6114, -122.3374

def haversine_km(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    a = np.sin((lat2 - lat1) / 2) ** 2 + np.cos(lat1) * np.cos(lat2) * np.sin((lon2 - lon1) / 2) ** 2
    return 2 * 6371 * np.arcsin(np.sqrt(a))

df_fe["DistanceToDowntown_km"] = haversine_km(df_fe["Latitude"], df_fe["Longitude"], DOWNTOWN_LAT, DOWNTOWN_LON)
df_fe["IsDowntown"] = (df_fe["Neighborhood"] == "DOWNTOWN").astype(int)

new_features.update({"DistanceToDowntown_km": "Localisation", "IsDowntown": "Localisation"})
df_fe.groupby("Neighborhood")["DistanceToDowntown_km"].median().sort_values().round(1).to_frame("Distance médiane (km)").T
```

Contrôle de cohérence : Downtown est à moins d'un kilomètre du point de référence et les quartiers du nord (North, Northwest) sont les plus éloignés, à environ 11 km. La feature est donc correctement calculée.

#### 5. Sources d'énergie présentes (information structurelle)

Les **quantités** consommées par source sont interdites (fuite). En revanche, savoir si le bâtiment est **raccordé** au gaz ou au réseau de vapeur urbain est une caractéristique de son équipement, connue indépendamment de la mesure de consommation. On ne garde que la présence (`> 0`), jamais la valeur.

- `HasNaturalGas` : le bâtiment utilise du gaz (chauffage, cuisine) ;
- `HasSteam` : raccordé au réseau de vapeur urbain (surtout le centre-ville).

L'électricité n'est pas retenue : 99,9 % des bâtiments en consomment, donc la variable serait quasi constante et n'apporterait aucune information.

```python
df_fe["HasNaturalGas"] = (df_fe["NaturalGas(kBtu)"] > 0).astype(int)
df_fe["HasSteam"] = (df_fe["SteamUse(kBtu)"] > 0).astype(int)
print(f"Part des bâtiments avec électricité : {(df_fe['Electricity(kBtu)'] > 0).mean():.1%} -> non retenue (quasi constante)")

new_features.update({"HasNaturalGas": "Sources d'énergie", "HasSteam": "Sources d'énergie"})
df_fe[["HasNaturalGas", "HasSteam"]].mean().map("{:.0%}".format).to_frame("Part des bâtiments")
```

#### 6. Contrôles finaux
On vérifie que les nouvelles features ne contiennent ni valeur manquante, ni infini, et qu'aucune n'est une colonne de relevé énergétique.

```python
fe_cols = list(new_features)
num_fe = df_fe[fe_cols].select_dtypes("number").columns

checks_fe = pd.DataFrame({
    "Catégorie": pd.Series(new_features),
    "Type": df_fe[fe_cols].dtypes.astype(str),
    "Nb NaN": df_fe[fe_cols].isna().sum(),
    "Nb infinis": pd.Series({c: int(np.isinf(df_fe[c]).sum()) if c in num_fe else 0 for c in fe_cols}),
    "Nb valeurs distinctes": df_fe[fe_cols].nunique(),
})
assert checks_fe["Nb NaN"].sum() == 0 and checks_fe["Nb infinis"].sum() == 0
assert not set(fe_cols) & set(LEAKAGE_COLS), "Une feature est une colonne de relevé énergétique !"

print(f"{len(fe_cols)} features créées ({n_cols_initial} -> {df_fe.shape[1]} colonnes)")
checks_fe
```

#### 7. Premier regard sur l'apport des nouvelles features

Coup d'œil rapide sur le lien entre quelques nouvelles features et la consommation (en log). C'est une simple vérification de bon sens : la sélection finale des features se fera dans la partie suivante.

```python
corr_fe = (df_fe[[c for c in num_fe if df_fe[c].nunique() > 2] + ["log_SiteEnergyUse"]]
           .corr(method="spearman")["log_SiteEnergyUse"].drop("log_SiteEnergyUse").sort_values())

fig, axes = plt.subplots(1, 2, figsize=(14, 5), gridspec_kw={"width_ratios": [1, 1.2]})
axes[0].barh(corr_fe.index, corr_fe.values, color=[ORANGE if v < 0 else BLUE for v in corr_fe.values], height=0.6)
axes[0].axvline(0, color=GREY, linewidth=0.8)
axes[0].set_xlabel("Corrélation de Spearman avec log10(consommation)")
axes[0].set_title("Features numériques créées")

order_fam = df_fe.groupby("UseFamily")["log_SiteEnergyUse"].median().sort_values(ascending=False).index
sns.boxplot(data=df_fe, y="UseFamily", x="log_SiteEnergyUse", order=order_fam, color=BLUE,
            fliersize=2, linewidth=1, ax=axes[1], boxprops={"alpha": 0.8})
axes[1].set_xlabel("log10(Consommation, kBtu)"); axes[1].set_ylabel("")
axes[1].set_title("Consommation par famille d'usage")
plt.tight_layout(); plt.show()

bin_feats = [c for c in num_fe if df_fe[c].nunique() == 2]
gap = pd.DataFrame({
    "Part de bâtiments à 1": df_fe[bin_feats].mean(),
    "Écart de médiane log10 (1 vs 0)": [df_fe.loc[df_fe[c] == 1, "log_SiteEnergyUse"].median()
                                        - df_fe.loc[df_fe[c] == 0, "log_SiteEnergyUse"].median() for c in bin_feats],
}).sort_values("Écart de médiane log10 (1 vs 0)", ascending=False)
gap
```

#### 8. Synthèse du feature engineering

| Catégorie | Features créées | Idée |
|---|---|---|
| Temporalité | `BuildingAge`, `ConstructionPeriod` | Âge au relevé et grande époque de construction (tranches à bornes fixes) |
| Structure | `log_GFATotal`, `log_GFABuilding`, `GFAPerFloor`, `log_GFAPerFloor`, `GFAPerBuilding`, `ParkingShare`, `HasParking`, `IsMultiBuilding` | Taille (en log), forme du bâtiment, part de parking, campus |
| Usages | `NumUseTypes`, `IsMultiUse`, `LargestUseShare`, `SecondUseShare`, `ThirdUseShare`, `HighIntensityShare`, `HasDataCenter`, `HasFoodService`, `HasHealthcare`, `HasLab`, `HasRefrigeratedStorage`, `UseFamily` | Mixité des usages, poids des usages énergivores, usage principal compressé de 56 à 10 familles |
| Localisation | `DistanceToDowntown_km`, `IsDowntown` | Position résumée par rapport au centre-ville |
| Sources d'énergie | `HasNaturalGas`, `HasSteam` | Équipement du bâtiment (présence seulement, jamais la quantité) |

**Ce que montrent les premiers contrôles**
- Les features de **taille** restent de loin les plus liées à la consommation (Spearman ≈ 0,7). Parmi les nouvelles features, `ParkingShare` (0,33), `DistanceToDowntown_km` (−0,29), `HighIntensityShare` (0,28) et `log_GFAPerFloor` (0,27) apportent un signal propre.
- Les indicateurs d'**usages énergivores** (santé, laboratoire, data center, restauration) et le raccordement à la **vapeur** vont avec une consommation médiane 3 à 16 fois plus élevée (écart de 0,5 à 1,2 en log10). Attention : `HasHealthcare`, `HasLab` et `HasRefrigeratedStorage` ne concernent que 1 à 2 % des bâtiments. Leur effet est donc estimé sur très peu d'exemples. `HasRefrigeratedStorage` va même dans l'autre sens, car ces entrepôts frigorifiques sont plutôt petits : la consommation totale mélange taille et intensité.
- `HasNaturalGas` ne différencie presque pas les bâtiments (écart de médiane 0,08). Il pourra en revanche être utile pour la cible émissions de CO2.
- `UseFamily` sépare bien les profils avec seulement 10 modalités : santé, alimentaire/restauration et hébergement en haut, entrepôts et lieux de culte/loisirs en bas.

**Garanties contre la fuite de données**
- Aucune valeur de relevé énergétique n'est utilisée. Seule la présence de gaz ou de vapeur est conservée, comme information structurelle.
- Toutes les features sont calculées ligne par ligne, avec des règles fixées à l'avance. Les étapes qui apprennent sur les données (encodage, regroupement des modalités rares de `Neighborhood`, standardisation) seront faites dans un `Pipeline` ajusté sur le seul jeu d'entraînement.
- Certaines features sont redondantes entre elles (`log_GFATotal` / `log_GFABuilding`, `GFAPerFloor` / `log_GFAPerFloor`, `IsDowntown` / `DistanceToDowntown_km`). Le tri sera fait à l'étape suivante avec la matrice de corrélation, comme le demande le template.
