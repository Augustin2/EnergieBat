#### 2. Compréhension des colonnes

Avant toute analyse, on classe les 46 colonnes selon leur **rôle** dans le projet. C'est essentiel : les relevés de consommation ne seront **pas connus** pour les nouveaux bâtiments, ils ne peuvent donc pas servir de features (sinon *fuite de données*).

| Groupe | Colonnes | Utilisation |
|---|---|---|
| Identifiants / texte libre | `OSEBuildingID`, `PropertyName`, `Address`, `TaxParcelIdentificationNumber` | Aucune (identifiants) |
| Constantes | `DataYear` (=2016), `City` (=Seattle), `State` (=WA), `Comments` (vide) | Aucune |
| Localisation | `ZipCode`, `CouncilDistrictCode`, `Neighborhood`, `Latitude`, `Longitude` | Features candidates |
| Structure | `YearBuilt`, `NumberofBuildings`, `NumberofFloors`, `PropertyGFATotal`, `PropertyGFAParking`, `PropertyGFABuilding(s)` | Features candidates |
| Usage | `BuildingType`, `PrimaryPropertyType`, `ListOfAllPropertyUseTypes`, `Largest/Second/ThirdLargestPropertyUseType(GFA)` | Filtrage + features candidates |
| Qualité de la donnée | `DefaultData`, `ComplianceStatus`, `Outlier` | **Filtrage des bâtiments peu fiables** |
| Relevés énergétiques | `SiteEnergyUse(kBtu)`, `SiteEnergyUseWN(kBtu)`, `SiteEUI…`, `SourceEUI…`, `Electricity…`, `NaturalGas…`, `SteamUse(kBtu)` | **Cibles candidates** / jamais en feature |
| Émissions | `TotalGHGEmissions`, `GHGEmissionsIntensity` | Cibles candidates / jamais en feature |
| Score ENERGY STAR | `ENERGYSTARScore`, `YearsENERGYSTARCertified` | Calculé *à partir* de la consommation → fuite, exclu |

Rappels utiles :
- **GFA** = *Gross Floor Area*, surface de plancher brute en pieds carrés (sf).
- **EUI** = *Energy Use Intensity* = consommation / surface (kBtu/sf).
- **WN** = *Weather Normalized*, consommation corrigée des variations météo.
- **Site** = énergie consommée sur place ; **Source** = énergie primaire (incluant pertes de production/transport).

```python
# Taux de valeurs manquantes par colonne
missing = (building_consumption.isna().mean() * 100).sort_values(ascending=False)
missing = missing[missing > 0]

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(missing.index[::-1], missing.values[::-1], color=BLUE, height=0.6)
for y, v in enumerate(missing.values[::-1]):
    ax.text(v + 1, y, f"{v:.1f} %", va="center", fontsize=8, color=GREY)
ax.set_xlabel("% de valeurs manquantes")
ax.set_xlim(0, 110)
ax.set_title("Taux de valeurs manquantes par colonne (jeu complet)")
plt.tight_layout(); plt.show()
```

**Lecture des valeurs manquantes — elles ne sont pas toutes des « trous » :**

- `Comments` est **entièrement vide** → colonne supprimée.
- `Outlier` est vide dans 99 % des cas : un vide signifie simplement *« le bâtiment n'est pas signalé comme outlier »*. C'est une information, pas un manque.
- `SecondLargestPropertyUseType(GFA)` / `ThirdLargest…` sont vides quand le bâtiment **n'a pas** de 2ᵉ ou 3ᵉ usage. La valeur logique est donc « aucun usage » et une surface de **0**.
- `YearsENERGYSTARCertified` est vide pour les bâtiments jamais certifiés ; `ENERGYSTARScore` n'est calculé que pour certains types d'usages éligibles. Ces deux colonnes sont de toute façon dérivées de la consommation (fuite) → exclues.
- Les relevés énergétiques ont ~9 manquants : ces bâtiments sont inutilisables pour l'entraînement (cible inconnue).

#### 3. Restriction au périmètre : bâtiments non destinés à l'habitation

On met en place un **journal de filtrage** pour tracer le nombre de lignes avant/après chaque étape.

```python
filter_log = []

def log_step(label, df_before, df_after):
    filter_log.append({
        "Étape": label,
        "Lignes avant": len(df_before),
        "Lignes après": len(df_after),
        "Lignes retirées": len(df_before) - len(df_after),
    })
    print(f"{label:<55} {len(df_before):>5} -> {len(df_after):>5}  (-{len(df_before) - len(df_after)})")

pd.crosstab(building_consumption["PrimaryPropertyType"], building_consumption["BuildingType"])
```

Le croisement `BuildingType` × `PrimaryPropertyType` montre que :
- les trois catégories **`Multifamily LR/MR/HR`** regroupent l'habitat collectif → à exclure ;
- quelques lignes classées *NonResidential* / *Campus* ont pourtant un usage principal **`Low-Rise Multifamily`** → exclues également ;
- `Residence Hall` (résidences étudiantes) est un usage **d'habitation** → exclu.

On conserve `Senior Care Community` : ce sont des établissements médicalisés (soins, restauration collective), dont le profil de consommation se rapproche davantage d'un bâtiment de services que d'un logement.

```python
df = building_consumption.copy()

before = df
df = df[~df["BuildingType"].str.startswith("Multifamily")]
log_step("BuildingType Multifamily (LR/MR/HR) exclus", before, df)

before = df
df = df[~df["PrimaryPropertyType"].isin(["Low-Rise Multifamily", "Residence Hall"])]
log_step("Usage principal résidentiel exclu", before, df)

# Uniformisation de libellés en doublon
df["PrimaryPropertyType"] = df["PrimaryPropertyType"].replace({"Office": "Small- and Mid-Sized Office"})
df["Neighborhood"] = df["Neighborhood"].str.upper().replace({"DELRIDGE NEIGHBORHOODS": "DELRIDGE"})

df["BuildingType"].value_counts()
```

Deux harmonisations de libellés au passage :
- les 3 bâtiments `Office` sont rattachés à `Small- and Mid-Sized Office` (catégorie la plus proche) ;
- `Neighborhood` mélangeait majuscules/minuscules (`North` vs `NORTH`) → passage en majuscules.

#### 4. Nettoyage : bâtiments aberrants, non conformes et incohérents

##### 4.1 Colonnes de qualité fournies par la ville
- `Outlier` : la ville a elle-même signalé certains bâtiments comme valeurs aberrantes (haute ou basse).
- `ComplianceStatus` : seuls les relevés **`Compliant`** sont validés. `Error - Correct Default Data` correspond exactement aux lignes `DefaultData = True` (valeurs par défaut saisies à la place des vraies) ; `Non-Compliant` et `Missing Data` sont non fiables.

```python
print(pd.crosstab(df["ComplianceStatus"], df["DefaultData"]), "\n")
print(df["Outlier"].value_counts(dropna=False))
```

```python
before = df
df = df[df["Outlier"].isna()]
log_step("Bâtiments signalés Outlier exclus", before, df)

before = df
df = df[df["ComplianceStatus"] == "Compliant"]
log_step("Relevés non conformes / données par défaut exclus", before, df)
```

##### 4.2 Cohérence des valeurs de features

On recherche des valeurs **physiquement impossibles ou contradictoires** :

```python
use_gfa_sum = df[["LargestPropertyUseTypeGFA", "SecondLargestPropertyUseTypeGFA",
                  "ThirdLargestPropertyUseTypeGFA"]].fillna(0).sum(axis=1)

checks = {
    "Consommation totale manquante": df["SiteEnergyUse(kBtu)"].isna(),
    "Consommation totale <= 0": df["SiteEnergyUse(kBtu)"] <= 0,
    "Émissions GES <= 0": df["TotalGHGEmissions"] <= 0,
    "Électricité négative": df["Electricity(kBtu)"] < 0,
    "Nombre de bâtiments = 0": df["NumberofBuildings"] == 0,
    "Nombre d'étages = 0": df["NumberofFloors"] == 0,
    "Surfaces des usages > 2x surface totale": use_gfa_sum > 2 * df["PropertyGFATotal"],
    "Somme GFA bâtiment + parking != GFA totale":
        (df["PropertyGFABuilding(s)"] + df["PropertyGFAParking"]) != df["PropertyGFATotal"],
}
pd.Series({k: int(v.sum()) for k, v in checks.items()}, name="Nb bâtiments").to_frame()
```

```python
# Le nombre d'étages : quelques valeurs extrêmes à examiner
df.nlargest(6, "NumberofFloors")[["PropertyName", "PrimaryPropertyType", "NumberofFloors", "PropertyGFATotal"]]
```

Une église de **99 étages** pour ~20 000 sf est manifestement une erreur de saisie (les autres valeurs élevées sont de vraies tours de bureaux du centre-ville, comme la Columbia Center à 76 étages).

**Décisions :**
- **Suppression** : cible manquante ou ≤ 0, émissions ≤ 0, électricité négative, l'église à 99 étages, et les bâtiments dont la somme des surfaces d'usage dépasse 2× la surface totale (surfaces contradictoires).
- **Correction plutôt que suppression** (pour ne pas perdre trop de bâtiments) :
  - `NumberofBuildings = 0` : une propriété déclarée possède forcément au moins un bâtiment → remplacé par **1** ;
  - `NumberofFloors = 0` : surtout des entrepôts/bâtiments bas ; remplacé par **1** (plain-pied).

```python
before = df
df = df[df["SiteEnergyUse(kBtu)"].notna() & (df["SiteEnergyUse(kBtu)"] > 0)]
log_step("Consommation manquante ou nulle exclue", before, df)

before = df
df = df[(df["TotalGHGEmissions"] > 0) & (df["Electricity(kBtu)"] >= 0)]
log_step("Émissions <= 0 / électricité négative exclues", before, df)

before = df
df = df[~((df["PrimaryPropertyType"] == "Worship Facility") & (df["NumberofFloors"] > 50))]
log_step("Nombre d'étages aberrant exclu", before, df)

before = df
use_gfa_sum = df[["LargestPropertyUseTypeGFA", "SecondLargestPropertyUseTypeGFA",
                  "ThirdLargestPropertyUseTypeGFA"]].fillna(0).sum(axis=1)
df = df[use_gfa_sum <= 2 * df["PropertyGFATotal"]]
log_step("Surfaces d'usage incohérentes exclues", before, df)

# Corrections
n_bld0, n_fl0 = (df["NumberofBuildings"] == 0).sum(), (df["NumberofFloors"] == 0).sum()
df.loc[df["NumberofBuildings"] == 0, "NumberofBuildings"] = 1
df.loc[df["NumberofFloors"] == 0, "NumberofFloors"] = 1
print(f"\nCorrigés : NumberofBuildings 0->1 : {n_bld0} | NumberofFloors 0->1 : {n_fl0}")
```

#### 5. Choix de la variable cible

Plusieurs colonnes peuvent servir de cible :

| Candidate | Pour | Contre |
|---|---|---|
| `SiteEnergyUse(kBtu)` | Consommation **totale** réelle : c'est exactement ce que demande la ville | Très asymétrique |
| `SiteEnergyUseWN(kBtu)` | Corrigée de la météo | Valeurs à 0 pour des bâtiments qui consomment (correction non calculée) |
| `SiteEUI(kBtu/sf)` | Indépendante de la taille | Intensité, pas consommation totale ; la ville veut la consommation |
| `TotalGHGEmissions` | Objectif carbone | Dépend surtout du **mix énergétique** (gaz, vapeur), non connu à partir des seules données structurelles |

```python
targets = ["SiteEnergyUse(kBtu)", "SiteEnergyUseWN(kBtu)", "SiteEUI(kBtu/sf)", "TotalGHGEmissions"]
print("Valeurs nulles par candidate :")
print((df[targets] == 0).sum(), "\n")
print("Corrélation (Spearman) entre candidates :")
df[targets].corr(method="spearman")
```

✅ **Cible retenue pour toute la suite du projet : `SiteEnergyUse(kBtu)`** — consommation totale d'énergie du site.

- Elle correspond directement à la demande métier (« prédire la consommation totale d'énergie ») ;
- elle n'a plus de valeur nulle ou manquante après nettoyage, contrairement à la version WN ;
- elle est fortement liée aux émissions (Spearman élevé), donc une bonne prédiction de consommation éclaire aussi l'enjeu carbone.

Les autres relevés énergétiques (EUI, électricité, gaz, vapeur, émissions, ENERGY STAR) seront **exclus des features** pour éviter toute fuite de données.

```python
TARGET = "SiteEnergyUse(kBtu)"

fig, axes = plt.subplots(1, 2, figsize=(11, 4))
axes[0].hist(df[TARGET] / 1e6, bins=60, color=BLUE)
axes[0].set_title("Consommation totale (échelle linéaire)")
axes[0].set_xlabel("SiteEnergyUse (millions de kBtu)"); axes[0].set_ylabel("Nb bâtiments")

axes[1].hist(np.log10(df[TARGET]), bins=40, color=BLUE)
axes[1].set_title("Consommation totale (échelle log10)")
axes[1].set_xlabel("log10(SiteEnergyUse en kBtu)")
plt.tight_layout(); plt.show()

print(f"Skewness brute : {df[TARGET].skew():.1f} | skewness log : {np.log10(df[TARGET]).skew():.2f}")
```

La distribution brute est **extrêmement asymétrique** (quelques très gros bâtiments : hôpitaux, université, tours). En échelle logarithmique, elle devient proche d'une **gaussienne**.
→ Pour la modélisation, il sera pertinent de prédire **log(cible)** (ou d'utiliser une transformation équivalente), pour que l'erreur ne soit pas dominée par une poignée de très gros bâtiments.

```python
df["log_SiteEnergyUse"] = np.log10(df[TARGET])
```

##### 7.1 Quanti × quanti

```python
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))
axes[0].scatter(np.log10(df["PropertyGFATotal"]), df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[0].set_xlabel("log10(Surface totale, sf)"); axes[0].set_ylabel("log10(Consommation, kBtu)")
axes[0].set_title("Consommation vs surface")

axes[1].scatter(df["NumberofFloors"], df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[1].set_xlabel("Nombre d'étages"); axes[1].set_title("Consommation vs nombre d'étages")

axes[2].scatter(df["YearBuilt"], df["log_SiteEnergyUse"], s=10, alpha=0.4, color=BLUE)
axes[2].set_xlabel("Année de construction"); axes[2].set_title("Consommation vs année de construction")
plt.tight_layout(); plt.show()
```

```python
corr_cols = ["log_SiteEnergyUse", "PropertyGFATotal", "PropertyGFABuilding(s)", "PropertyGFAParking",
             "LargestPropertyUseTypeGFA", "NumberofFloors", "NumberofBuildings", "BuildingAge",
             "Latitude", "Longitude"]
corr = df[corr_cols].corr(method="spearman")

fig, ax = plt.subplots(figsize=(9, 7))
mask = np.triu(np.ones_like(corr, dtype=bool), k=1)
sns.heatmap(corr, mask=mask, annot=True, fmt=".2f", cmap="RdBu_r", center=0, vmin=-1, vmax=1,
            annot_kws={"size": 8}, linewidths=1, linecolor="white", cbar_kws={"shrink": 0.7}, ax=ax)
ax.set_title("Corrélations de Spearman (features structurelles et log-cible)")
plt.tight_layout(); plt.show()
```

- La **surface** est de très loin le premier déterminant de la consommation (Spearman ≈ 0,74) : relation quasi linéaire en log-log.
- Les différentes surfaces sont **fortement corrélées entre elles** (multicolinéarité) : il faudra en choisir une ou deux pour les modèles linéaires.
- Le nombre d'étages est modérément corrélé (0,45), surtout via la taille.
- L'**âge** est négativement corrélé (−0,28) : les bâtiments récents (années 1990-2010) consomment davantage, mais ils sont aussi plus grands (corrélation âge/surface −0,28). C'est un effet de taille plus qu'un effet d'efficacité.
- La **latitude/longitude** brutes n'ont quasiment aucun lien linéaire avec la cible.

```python
# Pour séparer l'effet "taille" de l'effet "usage" : intensité énergétique (EUI) par usage
# (EUI utilisée ici uniquement pour l'analyse, pas comme feature)
order_eui = df.groupby("PrimaryPropertyType")["SiteEUI(kBtu/sf)"].median().sort_values(ascending=False).index
fig, ax = plt.subplots(figsize=(10, 7))
sns.boxplot(data=df, y="PrimaryPropertyType", x="SiteEUI(kBtu/sf)", order=order_eui,
            color=ORANGE, fliersize=2, linewidth=1, ax=ax, boxprops={"alpha": 0.8})
ax.set_xscale("log"); ax.set_xlabel("Intensité énergétique SiteEUI (kBtu/sf, échelle log)"); ax.set_ylabel("")
ax.set_title("Intensité énergétique par m² selon l'usage principal")
ax.tick_params(axis="y", labelsize=8)
plt.tight_layout(); plt.show()
```

- L'**usage** influence fortement la consommation, même à surface égale : hôpitaux, laboratoires, supermarchés et restaurants consomment **5 à 15 fois plus par unité de surface** (médiane ≈ 220-250 kBtu/sf contre 15-30) que les entrepôts, self-storage ou lieux de culte.
- La consommation totale élevée des universités et grands bureaux s'explique surtout par leur **taille**, alors que celle des hôpitaux, supermarchés et laboratoires s'explique par leur **intensité**.
→ L'usage principal (et la répartition des surfaces par usage) sera une feature clé, complémentaire de la surface.

Usages et quartiers ne sont pas indépendants : les **entrepôts** sont concentrés dans **Greater Duwamish**, les **grands bureaux et hôtels** dans **Downtown**, les écoles dans les quartiers résidentiels. Cela confirme que l'effet « quartier » observé plus haut passe en grande partie par le type de bâtiment.

##### 7.4 Mix énergétique (contexte, non utilisable en feature)

```python
mix = df[["Electricity(kBtu)", "NaturalGas(kBtu)", "SteamUse(kBtu)"]].div(df[TARGET], axis=0)
mix.columns = ["Électricité", "Gaz naturel", "Vapeur"]
share_by_type = mix.groupby(df["PrimaryPropertyType"]).mean().loc[order_eui] * 100

fig, ax = plt.subplots(figsize=(10, 7))
left = np.zeros(len(share_by_type))
for col, color in zip(share_by_type.columns, [BLUE, ORANGE, "#1baf7a"]):
    ax.barh(share_by_type.index, share_by_type[col], left=left, color=color, label=col,
            height=0.65, edgecolor="white", linewidth=1)
    left += share_by_type[col].values
ax.invert_yaxis(); ax.set_xlabel("Part moyenne de la consommation (%)")
ax.set_title("Mix énergétique moyen par usage principal")
ax.legend(loc="lower center", bbox_to_anchor=(0.5, 1.04), ncol=3, frameon=False); ax.tick_params(axis="y", labelsize=8)
plt.tight_layout(); plt.show()
```

L'électricité domine dans la plupart des usages, mais le **gaz** représente plus de la moitié de la consommation des écoles, lieux de culte, restaurants et maisons de retraite (chauffage, cuisine). La **vapeur** du réseau urbain reste marginale et concerne surtout hôpitaux, hôtels et laboratoires. #### 8. Traitement des valeurs manquantes restantes

On ne supprime **pas** les lignes ayant une valeur manquante : on interprète chaque cas.

```python
miss = df.isna().sum()
miss[miss > 0].sort_values(ascending=False).to_frame("Nb manquants")
```

```python
# 1) 2e et 3e usages absents = pas d'usage secondaire -> catégorie "None" et surface 0
for rank in ["Second", "Third"]:
    df[f"{rank}LargestPropertyUseType"] = df[f"{rank}LargestPropertyUseType"].fillna("None")
    df[f"{rank}LargestPropertyUseTypeGFA"] = df[f"{rank}LargestPropertyUseTypeGFA"].fillna(0)

# 2) Usage principal manquant : on reprend PrimaryPropertyType et la surface des bâtiments
mask_lu = df["LargestPropertyUseType"].isna()
df.loc[mask_lu, "LargestPropertyUseType"] = df.loc[mask_lu, "PrimaryPropertyType"]
df.loc[mask_lu, "LargestPropertyUseTypeGFA"] = df.loc[mask_lu, "PropertyGFABuilding(s)"]
df["ListOfAllPropertyUseTypes"] = df["ListOfAllPropertyUseTypes"].fillna(df["LargestPropertyUseType"])
print(f"Usage principal reconstitué pour {mask_lu.sum()} bâtiments")

# 3) Colonnes sans utilité ou constituant une fuite : suppression
cols_to_drop = ["Comments", "Outlier", "DefaultData", "ComplianceStatus", "DataYear", "City", "State",
                "YearsENERGYSTARCertified", "ENERGYSTARScore", "TaxParcelIdentificationNumber"]
df = df.drop(columns=cols_to_drop)

# ZipCode manquant : non bloquant (Latitude/Longitude/Neighborhood sont complets)
miss = df.isna().sum()
print("Manquants restants :", miss[miss > 0].to_dict())
print("Dimensions finales :", df.shape)
```

| Colonne | Raison des manquants | Traitement |
|---|---|---|
| `Second/ThirdLargestPropertyUseType(GFA)` | Pas de 2ᵉ/3ᵉ usage | `"None"` et surface 0 |
| `LargestPropertyUseType(GFA)` | Non renseigné pour quelques bâtiments mono-usage | Repris de `PrimaryPropertyType` / `PropertyGFABuilding(s)` |
| `ENERGYSTARScore`, `YearsENERGYSTARCertified` | Non éligible / non certifié | Supprimées (dérivées de la consommation → fuite) |
| `Comments`, `Outlier`, `DefaultData`, `ComplianceStatus` | Vides ou déjà exploitées pour le filtrage | Supprimées |
| `ZipCode` | Quelques adresses incomplètes | Conservé tel quel ; localisation couverte par lat/long et quartier |

**Périmètre et qualité**
- Sur 3 376 relevés, environ la moitié concerne de l'habitat collectif, hors périmètre. Après restriction au non résidentiel et retrait des bâtiments signalés outliers, non conformes ou incohérents, on conserve **1 509 bâtiments**, soit environ 90 % du périmètre non résidentiel brut (1 668).
- Les valeurs manquantes sont majoritairement **structurelles** (pas de 2ᵉ usage, non éligible ENERGY STAR) et ont été comblées logiquement plutôt que supprimées.

**Cible**
- `SiteEnergyUse(kBtu)` est retenue. Sa distribution est très asymétrique et **log-normale** → modéliser le log de la cible.

**Facteurs de consommation**
1. **La surface** est le facteur dominant (relation quasi linéaire en log-log).
2. **L'usage** module fortement l'intensité : hôpitaux, laboratoires, supermarchés et restaurants consomment beaucoup plus par m² que les entrepôts ou lieux de culte.
3. **La localisation** joue surtout via la composition du parc (Downtown = grands bureaux/hôtels) ; l'effet de l'**âge** passe surtout par la taille des bâtiments récents.

**Points d'attention pour la modélisation**
- Exclure toutes les mesures énergétiques et le score ENERGY STAR des features (fuite de données).
- Gérer la **multicolinéarité** des surfaces et les **catégories rares** d'usage.
- Exploiter les surfaces par usage (1ᵉʳ/2ᵉ/3ᵉ) pour décrire les bâtiments multi-usages.
